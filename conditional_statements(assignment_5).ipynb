```python
'''1. Positive, Negative, or Zero (if-elif-else)
Write a Python program that takes an integer from the user and checks whether the number is
Positive, Negative, or Zero.'''
user_input= int(input("Enter any number: "))
print(f"{user_input}")
if (user_input >0):
    print(f"{user_input} is a positive number.")
elif (user_input)==0:
    print(f"{user_input} is a zero.")
else:
    print(f"{user_input} is a negative number.")
```

```python
'''2. Even or Odd (if-else)
Write a Python program that asks the user to enter an integer and determines whether it is Even or
Odd.'''

num= int(input("Enter any number: "))
if num%2==0:
    print("Even Number")
else:
    print("Odd Number")
```

```python
'''3. Voting Eligibility (if-else)
Write a Python program that asks the user to enter their age. If the age is 18 or above, print 'You are eligible to vote.' Otherwise, print 'You are not eligible to vote.
'''

age = int(input("Enter your age: "))


if age >=18:
    print('You are eligible to vote.')
else:
    print('You are not eligible to vote.')
```

```python
'''4. Grade Calculator (if-elif-else)
Write a Python program that asks the user to enter their marks (0–100) and displays the grade:
90–100:A, 80–89:B, 70–79:C, 60–69:D, Below 60:F. '''

marks = float(input("Enter your marks"))

if marks < 0 or marks > 100:
    print("Invalid marks! Please enter marks between 0 and 100.")
elif marks >=90:
    print("A")
elif marks >=80 :
    print("B")
elif marks >=70 :
    print("C")
elif marks >=60:
    print('D')
else:
    ("F")

```

```python
'''5. Largest of Three Numbers
Write a Python program that takes three numbers as input from the user and prints the largest
number.'''

num1, num2, num3 = map(float, input("Enter three numbers: ").split())
print(num1)
print(num2)
print(num3)
if num1>num2 and num1>num3:
    print(f"Largest Number: {num1}")
elif num2>num1 and num2>num3:
    print(f"Largest Number {num2}")
else:
    print(f"Largest Number {num3}")
```
