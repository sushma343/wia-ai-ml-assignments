1. Create a 1D NumPy array containing the numbers from 10 to 50 (inclusive) with a step of 5.
● Print the entire array.
● Access and print the 3rd element.
● Access and print the last element using negative indexing.

```python
import numpy as np
```

```python
a1= np.arange(10,50,2)
print(a1)

print("3rd element:",a1[2])
print("Last Element:",a1[-1])
```

3. Create a 3D NumPy array of shape (2, 3, 4) filled with sequential numbers starting from
1.
● Print the entire array.
● Access and print the element at depth 0, row 1, column 2.


```python
a3=np.arange(1,25).reshape(2,3,4)
print(a3)

print("\nThe element at depth 0, row 1, column 2: ",a3[0][0][1])
```

4. Calculate mean median Variance, Standard deviation of
a. 12,32,56,32,87,65,45,32,12,32
b. 34,56,87,98,12,43,21,45,67,87,22,43

```python
#a

a = np.array([12, 32, 56, 32, 87, 65, 45, 32, 12, 32])

print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Variance:", np.var(a))
print("Standard Deviation:", np.std(a))
```

```python
b=np.array([34,56,87,98,12,43,21,45,67,87,22,43])
print("Mean: ",np.mean(b))
print("Median: ",np.median(b))
print("Variance: ",np.var(b))
print("Standard Deviation:",np.std(b))
```
