Q1. Create a class Engine with horsepower and a start() method.
Create a Car class that uses Composition (contains an Engine object).
Add a start_car() method.

#car has an enginer

```python
class Engine:
    def __init__(self,horsepower):
        self.horsepower=horsepower

    def start(self):
        print(f"Engine  with {self.horsepower}HP started.")

class Car:
    def __init__(self,name,engine_obj):
        self.name=name
        self.engine_obj=engine_obj

    def start_car(self):
        self.engine_obj.start()
        print(f"{self.name}  is started with {self.engine_obj.horsepower}HP engine ")

e1 = Engine(1000)
c1 = Car("Bugatti", e1)
c1.start_car()
```

Q2. Abstract class Notifier(ABC) with send(message, recipient)
• Implement EmailNotifier, SMSNotifier, PushNotifier
• User class that contains multiple Notifiers (Composition)
• notify_user(message) method that uses all attached notifiers

####

Abstraction (using ABC)
Composition (a User has multiple Notifier objects)
###
Create EmailNotifier

Create SMSNotifier

Create PushNotifier

↓

Put them inside User

↓

User.notify_user("Hello")

↓

Loop through every notifier

↓

Each notifier sends the message

```python
from abc import ABC, abstractmethod


# Abstract Class
class Notifier(ABC):

    @abstractmethod
    def send(self, message, recipient):
        pass


# Child Class 1
class EmailNotifier(Notifier):

    def send(self, message, recipient):
        print(f"Email sent to {recipient}: {message}")


# Child Class 2
class SMSNotifier(Notifier):

    def send(self, message, recipient):
        print(f"SMS sent to {recipient}: {message}")


# Child Class 3
class PushNotifier(Notifier):

    def send(self, message, recipient):
        print(f"Push notification sent to {recipient}: {message}")


# Composition
class User:

    def __init__(self, name, recipient):
        self.name = name
        self.recipient = recipient
        self.notifiers = []      #list to store notifier objects

    def add_notifier(self, notifier):
        self.notifiers.append(notifier)

    def notify_user(self, message):
        print(f"\nSending notifications to {self.name}...\n")

        for notifier in self.notifiers:
            notifier.send(message, self.recipient)



# Create notifier objects
email = EmailNotifier()
sms = SMSNotifier()
push = PushNotifier()

# Create user
user1 = User("Sushma", "sushma@gmail.com")

# Attach notifiers to the user
user1.add_notifier(email)
user1.add_notifier(sms)
user1.add_notifier(push)

# Send notification
user1.notify_user("Your assignment has been submitted successfully.")
```

Q3. Create an abstract class PaymentProcessor(ABC) with an
abstract method process_payment(amount).
Implement CreditCardProcessor and PayPalProcessor.
Create a ShoppingCart class that contains a PaymentProcessor
(Composition).


```python
from abc import ABC, abstractmethod


# Abstract Class
class PaymentProcessor(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


# Child Class 1
class CreditCardProcessor(PaymentProcessor):

    def process_payment(self, amount):
        print(f"Credit Card payment of Rs. {amount} processed successfully.")


# Child Class 2
class PayPalProcessor(PaymentProcessor):

    def process_payment(self, amount):
        print(f"PayPal payment of Rs. {amount} processed successfully.")


# Composition
class ShoppingCart:

    def __init__(self, payment_processor):
        self.payment_processor = payment_processor

    def checkout(self, amount):
        print("Processing your payment...")
        self.payment_processor.process_payment(amount)
        print("Order placed successfully.")


# Using Credit Card
credit = CreditCardProcessor()
cart1 = ShoppingCart(credit)

cart1.checkout(2500)
print()

# Using PayPal
paypal = PayPalProcessor()
cart2 = ShoppingCart(paypal)

cart2.checkout(1800)
```
