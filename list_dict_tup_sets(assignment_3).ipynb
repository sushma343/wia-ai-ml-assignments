3. Given a = [1, 2, 3, 5, 6, 7, 8, 9] and b = [1, 3, 5, 7, 9], return a new
list containing only the elements that are unique across both lists (items present
in one list but not both).


```python
#set()
a = [1, 2, 3, 5, 6, 7, 8, 9] 
b = [1, 3, 5, 7, 9]
result = list(set(a) ^ set(b)) # '^' is the symmetric difference operator. It returns items that are in either set, but not in both.
result
```

```python
#Using list comprehensions
result = [x for x in a if x not in b] + [x for x in b if x not in a]
result
```

```python
result = []

for item in a:
    if item not in b:
        result.append(item)

for item in b:
    if item not in a:
        result.append(item)
result
```

5. Given store1 = {'apple': 5, 'banana': 3} and store2 = {'orange': 4,
'banana': 10}, merge them into a single dictionary where store2 overwrites
duplicate keys from store1.

```python
#existing_dict.update(add_dict)
store1 = {'apple': 5, 'banana': 3}
store2 = {'orange': 4, 'banana': 10}
store1.update(store2)
store1
```

```python
#Using dictionary unpacking

merged = {**store1, **store2}

print(merged)
```

```python
#Using the merge operator
merged = store1 | store2

print(merged)
```
