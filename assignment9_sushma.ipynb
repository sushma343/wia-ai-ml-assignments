```python
#1. Use map() and lambda to double all numbers in this list: [1, 2, 3,4, 5].

l= [1,2,3,4,5]
print(list(map(lambda x:x*2, l)))
```

```python
#2. Use map() and lambda to convert this list of strings to uppercase: "ap[ple", "banana", "cherry"].
 
strings = ["apple", "banana", "cherry"]
upper_case_strings= map(lambda x: x.upper(), strings)
print(list(upper_case_strings))
```

```python
#3. Write a recursive function to find the factorial of a number and trace the output in a copy as well.

def fact(n):
    if n==0:
        return 1
    return n * fact(n-1)
fact(5)
```

```python
def factorial(n):
    print(f"Calling factorial({n})")

    if n == 1:
        print(f"Base case reached: factorial(1) = 1")
        return 1

    result = n * factorial(n - 1)

    print(f"Returning: factorial({n}) = {result}")

    return result


answer = factorial(5)
print("\nFinal Answer =", answer)
```

```python
#4. Write a recursive function to find the length of a string and trace the output as well.

def calc_len(string):
    if string=="":
        print("Base case reached")
        return 0
    
    result=  1+ calc_len(string[1:])
    print(f"Returning length of '{string}' = {result}")
    return result
Final_length =calc_len("hippopotamus")
print(Final_length)
```
