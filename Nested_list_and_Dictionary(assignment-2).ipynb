Nested Dictionary

```python
students = {
    "S001": {
        "name": "Ram",
        "age": 21,
        "subjects": ["Math", "Science"]
    },
    "S002": {
        "name": "Sita",
        "age": 22,
        "subjects": ["English", "Computer"]
    },
    "S003": {
        "name": "Hari",
        "age": 20,
        "subjects": ["Physics", "Chemistry"]
    }
}

print(students)
```

```python
#accessing elements
print("Student S001:", students["S001"])

print("Name:", students["S001"]["name"])

print("Age:", students["S002"]["age"])

print("Second Subject:", students["S003"]["subjects"][1])
```

```python
#Modifying Elements
students["S001"]["age"] = 24

students["S002"]["subjects"][0] = "Mathematics"

print(students)
```

```python
#Adding Elements
students["S004"] = {
    "name": "Gita",
    "age": 19,
    "subjects": ["Biology", "Economics"]
}

print(students)
```

```python
students["S001"]["subjects"].append("Computer")

print(students)
```

```python
#Removing Elements
students["S003"]["subjects"].remove("Physics")

print(students)
```

```python
students.pop("S002")

print(students)
```

```python
for sid, info in students.items():
    print("Student ID:", sid)
    print("Name:", info["name"])
    print("Age:", info["age"])
    print("Subjects:", info["subjects"])
    print()
```

```python
library = {
    "Science": {
        "books": ["Physics", "Chemistry", "Biology"],
        "floor": 2
    },
    "Computer": {
        "books": ["Python", "Java", "C++"],
        "floor": 3
    }
}

print("Original Dictionary")
print(library)

print("\nAccessing Data")
print(library["Computer"]["books"][0])

library["Science"]["floor"] = 1

library["Computer"]["books"].append("Data Science")

library["Science"]["books"].remove("Chemistry")

library["Management"] = {
        "books": ["Marketing", "Finance"],
        "floor": 4
        }

print("\nUpdated Dictionary")
print(library)
```
