Take user input of name, roll no, marks(only one) of the student and save it to csv file. Use read mode to find the  highest mark student and prnt the highest mark student info

```python

try:
    with open("student_info.csv", "w", newline="") as f:

        fields = ["Name", "Roll No", "Marks"]

        writer = csv.DictWriter(f, fieldnames=fields)
        writer.writeheader()

        while True:

            try:
                print("\nEnter Student Details")

                name = input("Name: ").strip()

                if name == "":
                    raise ValueError("Name cannot be empty.")

                roll = int(input("Roll No: "))

                if roll <= 0:
                    raise ValueError("Roll number must be greater than 0.")

                marks = float(input("Marks: "))

                if marks < 0 or marks > 100:
                    raise ValueError("Marks must be between 0 and 100.")

                writer.writerow({
                    "Name": name,
                    "Roll No": roll,
                    "Marks": marks
                })

                choice = input("\nDo you want to add another student? (y/n): ").lower()

                if choice != "y":
                    break

            except ValueError as e:
                print("Error:", e)
                print("Please enter the student details again.")

    print("\nStudent details saved successfully!")

except PermissionError:
    print("Permission denied! Unable to write to the file.")

except OSError as e:
    print("File Error:", e)
```

```python
#append
try:
    with open("student_info.csv", "a", newline="") as f:

        fields = ["Name", "Roll No", "Marks"]

        writer = csv.DictWriter(f, fieldnames=fields)

        while True:

            try:
                print("\nEnter Student Details")

                name = input("Name: ").strip()

                if name == "":
                    raise ValueError("Name cannot be empty.")

                roll = int(input("Roll No: "))

                if roll <= 0:
                    raise ValueError("Roll number must be greater than 0.")

                marks = float(input("Marks: "))

                if marks < 0 or marks > 100:
                    raise ValueError("Marks must be between 0 and 100.")

                writer.writerow({
                    "Name": name,
                    "Roll No": roll,
                    "Marks": marks
                })

                choice = input("\nAdd another student? (y/n): ").lower()

                if choice != "y":
                    break

            except ValueError as e:
                print("Error:", e)
                print("Please enter the details again.")

    print("\nNew student details added successfully!")

except PermissionError:
    print("Permission denied! Unable to write to the file.")

except OSError as e:
    print("File Error:", e)
```

```python

highest_student=None
highest_marks=0
try:
    with open("student_info.csv", "r") as f:

        reader = csv.DictReader(f)

        highest_student = None
        highest_marks = -1

        for row in reader:
            marks = float(row["Marks"])

            if marks > highest_marks:
                highest_marks = marks
                highest_student = row

        if highest_student:
            print("\nStudent with Highest Marks\n")
            print("Name    :", highest_student["Name"])
            print("Roll No :", highest_student["Roll No"])
            print("Marks   :", highest_student["Marks"])
        else:
            print("No student records found.")

except FileNotFoundError:
    print("File not found.")

except ValueError:
    print("Invalid marks found in the file.")

except Exception as e:
    print("Error:", e)
```
