1. Create a Rectangle class with:
● Attributes: length, width
● Methods:
○ area()
○ perimeter()
- Create 3 Rectangle objects. One object displays area() and the other
object displays perimeter()

```python
class Rectangle:#class creation
    def __init__(self, length, width): #attributes
        self.length=length
        self.width=width
#methods area and perimeter
    def area(self):
        return self.length*self.width
    def perimeter(self):
        return 2*(self.length+self.width)

# Create 3 Rectangle objects
rect1 = Rectangle(10, 5)
rect2 = Rectangle(8, 4)
rect3 = Rectangle(6, 3)

print("Area of rect1 =", rect1.area())


print("Perimeter of rect1 =", rect1.perimeter())
```

2. Create a Person class with:
● Attributes: name, age
● Methods:
○ introduce() → prints "Hi, I am [name], and I am [age] years old."
Create 2 Person objects and call introduce() on both.

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def intro(self):
            return(f"Hi, I am {self.name}, and I am {self.age} years old.")
        
person1=Person("Ram",22)
person2=Person("Rohan",23)

print(person1.intro())
print(person2.intro())
```

3. Create a Book class with:
● Attributes: title, author
● Methods:
○ details() → prints "Book: [title] by [author]"
Create 2 Book objects and show their details.

```python
class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author

    def details(self):
        return(f"Book: {self.title} by {self.author}")

book1= Book("Karnali Blues", "Buddhisagar")
book2= Book("Kafka on the Shore", "Murakami")

print(book1.details())
print(book2.details())
```

4. Create a Circle class with:
● Attribute: radius
● Methods:
○ area() → returns area (use 3.14 for π)
○ circumference() → returns circumference
Create one circle with radius 5 and print area and circumference

```python
class Circle:
    def __init__(self,radius):
        self.radius=radius
       

    def area(self):
        return (3.14* self.radius **2)
    
    def circumference(self):
        return (2*3.14*self.radius)
    
obj1= Circle(5)
print(obj1.area())
print(obj1.circumference())
```
