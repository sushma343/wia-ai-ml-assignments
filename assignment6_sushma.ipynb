Question 1: Student Grade Calculator
Write a Python program that calculates a student's total marks, percentage, and final
letter grade
based on their scores in Math, Science, English, History, and Art.
Requirements:
1. Prompt the user to enter the marks for all five subjects individually.
2. Convert each input into an integer.
3. Calculate the total marks obtained (out of 500).
4. Calculate the overall percentage.
5. Assign a grade: A (90-100), B (80-89.99), C (70-79.99), D (60-69.99), F (below 60).
6. Display Total Marks, Percentage (2 decimal places), and Final Grade.

```python
math = int(input("Enter your marks in Maths: "))
while math<0 or math>100:
    print("Invalid marks!")
    math = int(input("Enter your marks in Maths: "))

science = int(input("Enter your marks in  Science: "))
while science<0 or science>100:
    print("Invalid Marks!")
    science = int(input("Enter your marks in  Science: "))

english = int(input("Enter your marks in English: "))
while english<0 or english>100:
    print("Invalid Marks!")
    english = int(input("Enter your marks in English: "))


history = int(input("Enter your marls in hisotry: "))
while history<0 or history>100:
    print("Invalid Marks!")
    history = int(input("Enter your marls in hisotry: "))

art = int(input("Enter your marks in art: "))
while art<0 or art>100:
    print("Invalid Marks!")
    art = int(input("Enter your marks in art: "))

total_marks_obtained = math+ science + english + history + art
print(f"Total Marks Obtained = {total_marks_obtained}")

total_marks = 500
overall_percentage= total_marks_obtained/total_marks*100 
print(f"Overall percentage = {overall_percentage: .2f}")

if overall_percentage >=90:
    grade = "A"

elif overall_percentage >=80:
    grade = "B"

elif overall_percentage >=70:
    grade = 'C'

elif overall_percentage >=60:
    grade= 'D'

else:
    grade='F'

print(f"Final Grade: {grade}")


```

Question 3: FizzBuzz
Write a Python program that prints numbers from 1 to 100.
- Print "Fizz" for multiples of 3.
- Print "Buzz" for multiples of 5.
- Print "FizzBuzz" for multiples of both.
- Otherwise, print the number.


```python
for i in range(1,101):
    if i%3==0 and i%5 ==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
   
    else:
        print(i)
```

Question 4: Multiplication Table
Write a Python program that asks the user to enter a number and prints its multiplication
table from 1 to 10, except for even numbers, using a loop.


```python
num = int(input("Enter a number: "))

for i in range(1,11):
    if i%2==0:
        continue
    else:
        print(f'{num} x {i} = {num*i}')
```

Question 5: Number Guessing Game
Write a Python program that stores the secret number as 25.
- Ask the user to guess the number.
- Keep asking until the correct number is entered.
- If the guess is too low, print "Too Low!"
- If the guess is too high, print "Too High!"
- When the user guesses correctly, print "Congratulations! You guessed the correct
number."

```python
secret_number = 25
guess = int(input("Enter your guess: "))

while guess!=secret_number:
    if guess > secret_number:
        print("Too High!")
    elif guess< secret_number:
        print("Too Low!")

    guess = int(input("Guess again: "))
    
print("Congratulations! You guessed the correct number.")
```
