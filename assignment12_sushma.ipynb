2. Create a class Rectangle with attributes length and width. Include methods to calculate area
and perimeter.


```python
class Rectangle:
    def __init__(self, length, width):
        self.length=length
        self.width=width

    def calc_area(self):
        Area= self.length * self.width
        return f"Area: {Area}"
    
    
    def calc_perimeter(self):
        Perimeter= 2* (self.length + self.width)
        return f"Perimeter: {Perimeter}"
    
rec1=Rectangle(2,4)
print(rec1.calc_area())
print(rec1.calc_perimeter())
```

3. Write a class Car with attributes make, model, year, and speed. Add methods accelerate()
(increase speed by 10) and brake() (decrease speed by 10).

```python
class Car:
    def __init__(self, make, model, year, speed):
        self.make = make
        self.model = model
        self.year = year
        self.speed = speed

    def accelerate(self):
        self.speed += 10

    def brake(self):
        self.speed -= 10



car1 = Car("Toyota", "Corolla", 2022, 40)
print("Brand:",car1.make)
print('Model:',car1.model)
print('Year:',car1.year)

print("Initial Speed:", car1.speed)

car1.accelerate()
print("After Accelerating:", car1.speed)

car1.brake()
print("After Braking:", car1.speed)
```

4. Create a class BankAccount with attributes account_number and balance. Include methods
deposit(amount) and withdraw(amount).


```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance!")
     


account1 = BankAccount("ACC101", 5000)

print("Initial Balance:", account1.balance)

account1.deposit(1000)
print("After Deposit:", account1.balance)

account1.withdraw(1500)
print("After Withdrawal:", account1.balance)
```

5. Define a class Student with attributes name, age, and grades (list). Add a method to calculate
average grade

```python
class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades

    def calculate_average(self):
        return sum(self.grades) / len(self.grades)



student1 = Student("Sushma", 23, [80, 90, 70, 60])

print("Name:", student1.name)
print("Average Grade:", student1.calculate_average())
```

6. Create a class Employee with attributes name, salary, and department. Add a method
give_raise(percentage).


```python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    # <!-- def give_raise(self, percentage):
    #     self.salary += self.salary * (percentage / 100) -->

    def give_raise(self, percentage):
        if percentage > 0:
            self.salary += self.salary * (percentage / 100)
        else:
            print("Invalid percentage")


emp1 = Employee("Sushma", 50000, "AI Research")

print("Original Salary:", emp1.salary)

emp1.give_raise(10)

print("New Salary:", emp1.salary)
```

7. Write a class Circle with attribute radius. Include methods to calculate area and
circumference. (Use 3.14 for π)

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius * self.radius

    def calculate_circumference(self):
        return 2 * 3.14 * self.radius


# Create object
circle1 = Circle(5)

print("Area:", circle1.calculate_area())
print("Circumference:", circle1.calculate_circumference())
```

9. Define a class Laptop with attributes brand, ram, and price. Add a method
discount(percentage) that returns the discounted price.

```python
class Laptop:
    def __init__(self, brand, ram, price):
        self.brand = brand
        self.ram = ram
        self.price = price

    def discount(self, percentage):
        discounted_price = self.price - (self.price * percentage / 100)
        return discounted_price

lap1 = Laptop("Dell", 16, 100000)

print("Original Price:", lap1.price)
print("Discounted Price:", lap1.discount(20))
```

10. Create a class Counter with an attribute count. Include methods increment(), decrement(),
and reset().

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

    def decrement(self):
        self.count -= 1

    def reset(self):
        self.count = 0


counter1 = Counter()

print("Initial Count:", counter1.count)

counter1.increment()
print("After Increment:", counter1.count)

counter1.increment()
print("After Increment:", counter1.count)

counter1.decrement()
print("After Decrement:", counter1.count)

counter1.reset()
print("After Reset:", counter1.count)
```
