1. Create a base class Shape :
__init__(self, name)
area(self) method (return 0 by default)
info(self) method that returns name only
Create a subclass Rectangle that inherits from Shape:
Add attributes: length and width
Override area() to calculate and return the rectangle area (length ×
width)
Override info() to display name, dimensions, and area

```python

class Shape:
    def __init__(self,name):
        self.name=name

    def area(self):
        return 0

    def info(self):
        return self.name


class Rectangle(Shape):
    def __init__(self, name, length, width):
        super().__init__(name)
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def info(self):
        return (
            f"Name: {self.name}\n"
            f"Length: {self.length}\n"
            f"Width: {self.width}\n"
            f"Area: {self.area()}"
        )


rect = Rectangle("Rectangle", 8, 4)


print(rect.info())
```

2. Create a parent class Employee:
__init__(self, name, salary)
calc_bonus(self) method calc bonus of 10% of salary and return it
display(self) to display employee name and salary.
Create Two subclass that inherits from Employee. SoftwareEngineer
and DevopsDeveloper.
Add Attributes - designation
Override calc_bonus method in two child class and calc bonus with
20% of salary for SoftwareEngineer and 15% of salary for
DevopsDeveloper and return the bonus information along with
employee information.


```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calc_bonus(self):
        return self.salary * 0.10

    def display(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")


class SoftwareEngineer(Employee):
    def __init__(self, name, salary, designation):
        super().__init__(name, salary)
        self.designation = designation

    def calc_bonus(self):
        return self.salary * 0.20

    def display(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")
        print(f"Designation: {self.designation}")
        print(f"Bonus: {self.calc_bonus()}")


class DevopsDeveloper(Employee):
    def __init__(self, name, salary, designation):
        super().__init__(name, salary)
        self.designation = designation

    def calc_bonus(self):
        return self.salary * 0.15

    def display(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")
        print(f"Designation: {self.designation}")
        print(f"Bonus: {self.calc_bonus()}")



se = SoftwareEngineer("Sushma", 80000, "Software Engineer")
dd = DevopsDeveloper("Aayush", 70000, "DevOps Developer")

print("Software Engineer")
se.display()

print("\nDevOps Developer")
dd.display()
```

3. Create two base classes:
Flyable with fly() method
Swimmable with swim() method
Create class AmphibiousVehicle that inherits from both Flyable and
Swimmable
Add name and speed attributes
Override both fly() and swim() methods with specific behavior
Add perform_actions() method that calls both overridden methods

```python
class Flyable:
    def fly(self):
        print("Flying")


class Swimmable:
    def swim(self):
        print("Swimming")


class AmphibiousVehicle(Flyable, Swimmable):
    def __init__(self, name, speed):
        self.name = name
        self.speed = speed

    def fly(self):
        print(f"{self.name} is flying at {self.speed} km/h.")

    def swim(self):
        print(f"{self.name} is swimming at {self.speed} km/h.")

    def perform_actions(self):
        self.fly()
        self.swim()



vehicle = AmphibiousVehicle("Sea Plane", 300)


vehicle.perform_actions()
```

HYBRID INHERITANCE

Hierarchical inheritance (Student and Employee inherit from Person)
Multiple inheritance (TeachingAssistant inherits from both Student and Employee)

```python
# Parent Class
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"My name is {self.name}.")


# Child Class 1
class Student(Person):
    def study(self):
        print(f"{self.name} is studying.")


# Child Class 2
class Employee(Person):
    def work(self):
        print(f"{self.name} is working.")


# Hybrid Inheritance
class TeachingAssistant(Student, Employee):
    def assist(self):
        print(f"{self.name} is assisting the teacher.")


# Create Object
ta = TeachingAssistant("Sushma")

ta.introduce()   # From Person
ta.study()       # From Student
ta.work()        # From Employee
ta.assist()      # Own method
```

Multiple Inheritance: C inherits from A and B.
Multilevel Inheritance: D inherits from C.

```python
# Parent Class A
class Camera:
    def take_photo(self):
        print("Taking a photo.")


# Parent Class B
class Phone:
    def make_call(self):
        print("Making a phone call.")


# Child Class C (Multiple Inheritance)
class SmartPhone(Camera, Phone):
    def browse_internet(self):
        print("Browsing the internet.")


# Child Class D (Multilevel Inheritance)
class GamingPhone(SmartPhone):
    def play_games(self):
        print("Playing high-end games.")


# Create object
phone = GamingPhone()

phone.take_photo()       # From Camera
phone.make_call()        # From Phone
phone.browse_internet()  # From SmartPhone
phone.play_games()       # From GamingPhone
```

Single inheritance
C inherits from A
D inherits from B
Multiple inheritance
E inherits from both C and D

```python
# Parent Class A
class Student:
    def study(self):
        print("Student is studying.")


# Parent Class B
class Athlete:
    def practice(self):
        print("Athlete is practicing.")


# Child Class C
class CollegeStudent(Student):
    def attend_class(self):
        print("Attending college classes.")


# Child Class D
class FootballPlayer(Athlete):
    def play_match(self):
        print("Playing a football match.")


# Hybrid Class E
class StudentAthlete(CollegeStudent, FootballPlayer):
    def introduce(self):
        print("I am a student athlete.")


# Create object
person = StudentAthlete()

person.study()
person.attend_class()
person.practice()
person.play_match()
person.introduce()
```
