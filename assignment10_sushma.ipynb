Student Grade Calculator
Write a Python program that calculates a student's total marks, percentage, and final
letter grade
based on their scores in Math, Science, English, History, and Art.
Requirements:
1. Prompt the user to enter the marks for all five subjects individually.
2. Convert each input into an integer.
3. Calculate the total marks obtained (out of 500).
4. Calculate the overall percentage.
5. Assign a grade: A (90-100), B (80-89.99), C (70-79.99), D (60-69.99), F (below 60).
6. Display Total Marks, Percentage (2 decimal places), and Final Grade.

```python

def calculate_grade(math, science, english, history, art):
    total = math + science + english + history + art
    percentage = total / 5

    if percentage >= 90:
        grade = "A"
    elif percentage >= 80:
        grade = "B"
    elif percentage >= 70:
        grade = "C"
    elif percentage >= 60:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade


try:
    math = float(input("Enter Math marks: "))
    science = float(input("Enter Science marks: "))
    english = float(input("Enter English marks: "))
    history = float(input("Enter History marks: "))
    art = float(input("Enter Art marks: "))

    # Validate marks
    marks = [math, science, english, history, art]

    for mark in marks:
        if mark < 0 or mark > 100:
            raise ValueError("Marks must be between 0 and 100.")

except ValueError as e:
    print(f"Error: {e}")

else:
    total, percentage, grade = calculate_grade(
        math, science, english, history, art
    )

    print("\n----- Student Result -----")
    print(f"Total Marks : {total}/500")
    print(f"Percentage  : {percentage:.2f}%")
    print(f"Grade       : {grade}")

finally:
    print("\nProgram finished.")
```

```python
#printing even numbers
def print_even_numbers(start, end):
    if start > end:
        raise ValueError("Starting number cannot be greater than ending number.")

    print("Even numbers are:")
    for i in range(start, end + 1):
        if i % 2 == 0:
            print(i)


try:
   
    print_even_numbers(1, 100)

except ValueError as e:
    print(f"Error: {e}")

finally:
    print("Program finished.")
```

```python
#multiplication table
def multiplication_table(number):
    print(f"\nMultiplication Table of {number} (Odd Multipliers Only):")

    for i in range(1, 11):
        if i % 2 == 0:
            continue      # Skip even numbers

        print(f"{number} x {i} = {number * i}")


try:
    number = int(input("Enter a number: "))

    multiplication_table(number)

except ValueError:
    print("Error: Please enter a valid whole number.")

finally:
    print("Program finished.")
```

```python
#Numbers Guessing Game
def guess_number():
    secret_number = 25

    while True:
        try:
            guess = int(input("Guess the secret number: "))

            if guess < secret_number:
                print("Too Low!")

            elif guess > secret_number:
                print("Too High!")

            else:
                print("Congratulations! You guessed the correct number.")
                break

        except ValueError:
            print("Error: Please enter a valid whole number.")

    print("Game Over.")


guess_number()
```
